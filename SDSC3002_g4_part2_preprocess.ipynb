```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler

data = pd.read_csv("imdb_top_1000.csv")

print(data['Released_Year'].unique())


data['Released_Year'] = pd.to_numeric(data['Released_Year'], errors='coerce') 
data['Released_Year'].fillna(data['Released_Year'].median(), inplace=True)  
data['Released_Year'] = data['Released_Year'].astype(int)  


data['Released_Decade'] = (data['Released_Year'] // 10) * 10
print(data[['Released_Year', 'Released_Decade']].head())


data['Gross'] = data['Gross'].str.replace(',', '').astype(float)  
data['Gross'].fillna(data['Gross'].median(), inplace=True)

data['Runtime'] = data['Runtime'].str.replace(' min', '').astype(int) 

data.drop(['Poster_Link', 'Overview'], axis=1, inplace=True)


data['Certificate'] = data['Certificate'].fillna('Unknown')  
data['Certificate'] = data['Certificate'].astype('category').cat.codes

data['is_popular'] = (data['IMDB_Rating'] > 8).astype(int) 
```

```python
import pandas as pd
import numpy as np



people = pd.concat([
    data['Director'],
    data['Star1'],
    data['Star2'], 
    data['Star3'],
    data['Star4']
]).unique()

result = pd.DataFrame({'Name': people})


metrics = []
for name in people:
 
    movies = data[(data['Director']==name) | 
                 (data['Star1']==name) |
                 (data['Star2']==name) |
                 (data['Star3']==name) |
                 (data['Star4']==name)]
    
    metrics.append({
        'IMDB': movies['IMDB_Rating'].mean(),
        'Meta': movies['Meta_score'].mean(),
        'Votes': movies['No_of_Votes'].mean(),
        'Gross': movies['Gross'].mean(),
        'Movie_Count': len(movies)
    })

metrics_df = pd.DataFrame(metrics)
result = pd.concat([result, metrics_df], axis=1)

for col in ['IMDB', 'Meta', 'Votes', 'Gross']:
    result[col] = result[col].fillna(result[col].mean())

result['IMDB'] = result['IMDB'].round(2)
result['Meta'] = result['Meta'].round(2)
result['Votes'] = result['Votes'].round(2)
result['Gross'] = result['Gross'].round(2)


result = result.sort_values('Name')


print(f"共统计 {len(result)} 位影视人员:")
print(result.head())


result.to_csv('people_stats.csv', index=False, float_format='%.2f')
```
